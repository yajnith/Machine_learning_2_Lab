```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("find_s_dataset.csv")

print("Dataset:")
print(data)

# Separate attributes and target
X = data.iloc[:, :-1]
y = data.iloc[:, -1]

# ---------------- FIND-S ALGORITHM ----------------

# Start with first positive example
hypothesis = None

for i in range(len(data)):
    if str(y.iloc[i]).lower() == "yes":
        hypothesis = list(X.iloc[i])
        break

# Generalize hypothesis using other positive examples
for i in range(len(data)):
    if str(y.iloc[i]).lower() == "yes":

        for j in range(len(hypothesis)):
            if hypothesis[j] != X.iloc[i, j]:
                hypothesis[j] = "?"

print("\nFinal Hypothesis:")
print(hypothesis)


# ---------------- GRAPH 1 ----------------

# Count positive and negative examples
positive = sum(y.astype(str).str.lower() == "yes")
negative = sum(y.astype(str).str.lower() == "no")

plt.figure(figsize=(6, 4))

plt.bar(
    ["Positive (Yes)", "Negative (No)"],
    [positive, negative]
)

plt.title("Positive vs Negative Examples")
plt.xlabel("Class")
plt.ylabel("Number of Examples")

plt.show()


# ---------------- GRAPH 2 ----------------

# Plot each attribute's number of unique values
unique_values = X.nunique()

plt.figure(figsize=(8, 5))

plt.bar(
    unique_values.index,
    unique_values.values
)

plt.title("Unique Values in Each Attribute")
plt.xlabel("Attributes")
plt.ylabel("Number of Unique Values")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
