```python
# ============================================================================
# BAGGING AND BOOSTING FOR SUPERVISED CLASSIFICATION
# BCI702 : Machine Learning
#
# Objective:
#   1. Read a supervised classification dataset
#   2. Split the dataset into training and testing sets
#   3. Apply Bagging
#   4. Apply Boosting
#   5. Compare model performance
#   6. Display accuracy, precision, recall, F1-score and ROC-AUC
#   7. Plot confusion matrices and ROC curves
# ============================================================================


# ============================================================================
# SECTION 1 : IMPORT LIBRARIES
# ============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    BaggingClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 110
plt.rcParams["font.size"] = 11


# ============================================================================
# SECTION 2 : READ THE SUPERVISED DATASET
# ============================================================================
#
# Breast Cancer Wisconsin dataset:
#
#   Features : 30 numerical medical measurements
#   Target   : malignant / benign
#
# Target encoding:
#
#   0 = malignant
#   1 = benign
# ============================================================================


data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="target"
)


print("=" * 80)
print(" SUPERVISED CLASSIFICATION DATASET ".center(80, "="))
print("=" * 80)

print("\nDataset shape :", X.shape)

print("\nNumber of features :", X.shape[1])

print("Number of samples  :", X.shape[0])

print("\nTarget classes:")

print(
    pd.Series(y)
    .value_counts()
    .rename(index={
        0: "Malignant",
        1: "Benign"
    })
)


# ============================================================================
# SECTION 3 : DISPLAY DATASET
# ============================================================================

print("\nFirst 10 rows:")

display(
    X.head(10)
)


print("\nTarget values:")

display(
    y.head(10)
)


# ============================================================================
# SECTION 4 : CHECK MISSING VALUES
# ============================================================================


print("\nMissing values:")

missing_values = X.isnull().sum().sum()

print(
    "Total missing values =",
    missing_values
)


# ============================================================================
# SECTION 5 : BASIC DATA VISUALIZATION
# ============================================================================


plt.figure(
    figsize=(7, 5)
)

sns.countplot(
    x=y.map({
        0: "Malignant",
        1: "Benign"
    })
)

plt.title(
    "Class Distribution"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Number of Samples"
)

plt.tight_layout()

plt.show()


# ============================================================================
# SECTION 6 : TRAIN / TEST SPLIT
# ============================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\nTraining samples :", len(X_train))

print("Testing samples  :", len(X_test))


# ============================================================================
# SECTION 7 : BAGGING MODEL
# ============================================================================
#
# Bagging = Bootstrap Aggregating
#
# Multiple decision trees are trained on different bootstrap samples.
#
# Final prediction is obtained by combining predictions from all trees.
# ============================================================================


base_tree = DecisionTreeClassifier(
    max_depth=None,
    random_state=42
)


bagging_model = BaggingClassifier(
    estimator=base_tree,
    n_estimators=100,
    max_samples=0.8,
    max_features=1.0,
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)


bagging_model.fit(
    X_train,
    y_train
)


bagging_pred = bagging_model.predict(
    X_test
)

bagging_prob = bagging_model.predict_proba(
    X_test
)[:, 1]


# ============================================================================
# SECTION 8 : BOOSTING MODEL - ADABOOST
# ============================================================================
#
# AdaBoost trains weak learners sequentially.
#
# Each new learner concentrates more on incorrectly classified examples.
# ============================================================================


boosting_model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(
        max_depth=1,
        random_state=42
    ),
    n_estimators=100,
    learning_rate=0.8,
    random_state=42
)


boosting_model.fit(
    X_train,
    y_train
)


boosting_pred = boosting_model.predict(
    X_test
)

boosting_prob = boosting_model.predict_proba(
    X_test
)[:, 1]


# ============================================================================
# SECTION 9 : SECOND BOOSTING MODEL - GRADIENT BOOSTING
# ============================================================================
#
# Gradient Boosting builds trees sequentially.
# Each tree attempts to reduce the errors made by the previous ensemble.
#
# This is included as an additional boosting comparison.
# ============================================================================


gradient_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=2,
    random_state=42
)


gradient_model.fit(
    X_train,
    y_train
)


gradient_pred = gradient_model.predict(
    X_test
)

gradient_prob = gradient_model.predict_proba(
    X_test
)[:, 1]


# ============================================================================
# SECTION 10 : PERFORMANCE FUNCTION
# ============================================================================


def evaluate_model(
    model_name,
    y_true,
    predictions,
    probabilities
):

    accuracy = accuracy_score(
        y_true,
        predictions
    )

    precision = precision_score(
        y_true,
        predictions
    )

    recall = recall_score(
        y_true,
        predictions
    )

    f1 = f1_score(
        y_true,
        predictions
    )

    auc = roc_auc_score(
        y_true,
        probabilities
    )

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": auc
    }


# ============================================================================
# SECTION 11 : EVALUATE ALL MODELS
# ============================================================================


results = []


results.append(
    evaluate_model(
        "Bagging",
        y_test,
        bagging_pred,
        bagging_prob
    )
)


results.append(
    evaluate_model(
        "AdaBoost",
        y_test,
        boosting_pred,
        boosting_prob
    )
)


results.append(
    evaluate_model(
        "Gradient Boosting",
        y_test,
        gradient_pred,
        gradient_prob
    )
)


results_df = pd.DataFrame(
    results
)


print("\n" + "=" * 80)

print(
    " MODEL PERFORMANCE COMPARISON ".center(80, "=")
)

print("=" * 80)


display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)


# ============================================================================
# SECTION 12 : CLASSIFICATION REPORTS
# ============================================================================


print("\n" + "=" * 80)

print(
    " BAGGING CLASSIFICATION REPORT ".center(80, "=")
)

print("=" * 80)

print(
    classification_report(
        y_test,
        bagging_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


print("\n" + "=" * 80)

print(
    " ADABOOST CLASSIFICATION REPORT ".center(80, "=")
)

print("=" * 80)

print(
    classification_report(
        y_test,
        boosting_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


print("\n" + "=" * 80)

print(
    " GRADIENT BOOSTING CLASSIFICATION REPORT ".center(80, "=")
)

print("=" * 80)

print(
    classification_report(
        y_test,
        gradient_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


# ============================================================================
# SECTION 13 : CONFUSION MATRICES
# ============================================================================


cm_bagging = confusion_matrix(
    y_test,
    bagging_pred
)

cm_adaboost = confusion_matrix(
    y_test,
    boosting_pred
)

cm_gradient = confusion_matrix(
    y_test,
    gradient_pred
)


fig, axes = plt.subplots(
    1,
    3,
    figsize=(16, 5)
)


sns.heatmap(
    cm_bagging,
    annot=True,
    fmt="d",
    cmap="Blues",
    ax=axes[0],
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

axes[0].set_title(
    "Bagging Confusion Matrix"
)

axes[0].set_xlabel(
    "Predicted"
)

axes[0].set_ylabel(
    "Actual"
)


sns.heatmap(
    cm_adaboost,
    annot=True,
    fmt="d",
    cmap="Greens",
    ax=axes[1],
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

axes[1].set_title(
    "AdaBoost Confusion Matrix"
)

axes[1].set_xlabel(
    "Predicted"
)

axes[1].set_ylabel(
    "Actual"
)


sns.heatmap(
    cm_gradient,
    annot=True,
    fmt="d",
    cmap="Oranges",
    ax=axes[2],
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

axes[2].set_title(
    "Gradient Boosting Confusion Matrix"
)

axes[2].set_xlabel(
    "Predicted"
)

axes[2].set_ylabel(
    "Actual"
)


plt.suptitle(
    "Confusion Matrix Comparison",
    fontsize=14
)

plt.tight_layout()

plt.show()


# ============================================================================
# SECTION 14 : ROC CURVES
# ============================================================================


fpr_bagging, tpr_bagging, _ = roc_curve(
    y_test,
    bagging_prob
)

fpr_adaboost, tpr_adaboost, _ = roc_curve(
    y_test,
    boosting_prob
)

fpr_gradient, tpr_gradient, _ = roc_curve(
    y_test,
    gradient_prob
)


auc_bagging = roc_auc_score(
    y_test,
    bagging_prob
)

auc_adaboost = roc_auc_score(
    y_test,
    boosting_prob
)

auc_gradient = roc_auc_score(
    y_test,
    gradient_prob
)


plt.figure(
    figsize=(9, 7)
)


plt.plot(
    fpr_bagging,
    tpr_bagging,
    linewidth=2,
    label="Bagging (AUC = %.3f)"
    % auc_bagging
)


plt.plot(
    fpr_adaboost,
    tpr_adaboost,
    linewidth=2,
    label="AdaBoost (AUC = %.3f)"
    % auc_adaboost
)


plt.plot(
    fpr_gradient,
    tpr_gradient,
    linewidth=2,
    label="Gradient Boosting (AUC = %.3f)"
    % auc_gradient
)


plt.plot(
    [0, 1],
    [0, 1],
    "k--",
    label="Random Classifier"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve Comparison"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================================
# SECTION 15 : PERFORMANCE BAR CHART
# ============================================================================


metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
]


plot_data = results_df.set_index(
    "Model"
)[metrics]


ax = plot_data.plot(
    kind="bar",
    figsize=(12, 6),
    width=0.8
)


plt.title(
    "Bagging vs Boosting Performance"
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Model"
)

plt.ylim(
    0.8,
    1.02
)

plt.xticks(
    rotation=0
)

plt.legend(
    loc="lower right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================================
# SECTION 16 : MODEL TRAINING PARAMETERS
# ============================================================================


print("\n" + "=" * 80)

print(
    " MODEL CONFIGURATION ".center(80, "=")
)

print("=" * 80)


print("""
BAGGING
-------
Base learner       : Decision Tree
Number of trees    : 100
Bootstrap sampling : True
Sample fraction    : 80%


ADABOOST
--------
Base learner       : Decision Tree
Tree depth         : 1
Number of learners : 100
Learning rate      : 0.8


GRADIENT BOOSTING
-----------------
Number of trees    : 100
Learning rate      : 0.05
Tree depth         : 2
""")


# ============================================================================
# SECTION 17 : FINAL AUTOMATED COMPARISON
# ============================================================================


print("\n" + "=" * 80)

print(
    " FINAL PERFORMANCE REPORT ".center(80, "=")
)

print("=" * 80)


for _, row in results_df.iterrows():

    print(
        "\n%s" % row["Model"]
    )

    print(
        "  Accuracy  : %.2f%%"
        % (row["Accuracy"] * 100)
    )

    print(
        "  Precision : %.2f%%"
        % (row["Precision"] * 100)
    )

    print(
        "  Recall    : %.2f%%"
        % (row["Recall"] * 100)
    )

    print(
        "  F1 Score  : %.2f%%"
        % (row["F1 Score"] * 100)
    )

    print(
        "  ROC-AUC   : %.4f"
        % row["ROC-AUC"]
    )


# ============================================================================
# SECTION 18 : EXAMPLE PREDICTIONS
# ============================================================================


print("\n" + "=" * 80)

print(
    " SAMPLE TEST PREDICTIONS ".center(80, "=")
)

print("=" * 80)


sample_indices = X_test.index[:10]


sample_results = pd.DataFrame({
    "Actual": y_test.loc[sample_indices].values,
    "Bagging": bagging_model.predict(
        X_test.loc[sample_indices]
    ),
    "AdaBoost": boosting_model.predict(
        X_test.loc[sample_indices]
    ),
    "GradientBoosting": gradient_model.predict(
        X_test.loc[sample_indices]
    )
})


sample_results = sample_results.replace({
    0: "Malignant",
    1: "Benign"
})


display(
    sample_results
)


# ============================================================================
# SECTION 19 : CONCLUSION
# ============================================================================


print("\n" + "=" * 80)

print(
    " CONCLUSION ".center(80, "=")
)

print("=" * 80)

print("""
Bagging and Boosting are ensemble learning techniques.

BAGGING:
- Creates multiple bootstrap samples from the training data.
- Trains several models independently.
- Combines their predictions.
- Primarily helps reduce variance and improve stability.

BOOSTING:
- Builds models sequentially.
- Each new model attempts to correct errors made by previous models.
- Can produce a strong classifier from weak learners.

In this experiment:
- Bagging uses an ensemble of decision trees.
- AdaBoost uses sequential weak decision trees.
- Gradient Boosting uses sequential trees that minimize prediction error.
- Accuracy, precision, recall, F1-score and ROC-AUC are used for comparison.

For the final report, compare all models using the generated
performance table rather than relying only on accuracy.
""")

print("=" * 80)
print(" END OF EXPERIMENT ".center(80, "="))
print("=" * 80)
```
